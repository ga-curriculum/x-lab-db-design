```python
import numpy as np
import pandas as pd
import sqlite3
from faker import Faker
import hashlib
```

```python
f = Faker()

def to_id(s):
    return hashlib.sha256(s.encode('utf8')).hexdigest()[-16:]
```

### Users

```python
f.seed_instance(1)

N_USERS = 200

users = pd.DataFrame({
    'first': [f.first_name() for _ in range(N_USERS)],
    'last': [f.last_name() for _ in range(N_USERS)]
})

users['email'] = (users['first'] + '.' + users['last'] + "@gmail.com").str.lower()
users['user_id'] = users.email.apply(to_id)
```

### Payment Methods

```python
f.seed_instance(2)

N_PAY = 250

payment_methods = pd.DataFrame({
    'cc_number': [f.credit_card_number() for _ in range(N_PAY)],
    'cc_expire': [f.credit_card_expire() for _ in range(N_PAY)],
    'cc_security': [f.credit_card_security_code() for _ in range(N_PAY)],
})

user_ids = users.user_id.unique().tolist()

payment_methods['user_id'] = np.append(
    np.random.choice(user_ids, N_USERS, replace=False),
    np.random.choice(user_ids, N_PAY - N_USERS, replace=True),
)
```

```python
class_names = [
    'warrior', 'mage', 'rogue', 'paladin', 'ranger',
    'cleric', 'druid', 'necromancer', 'bard', 'warlock'
]
```

```python
np.random.seed(789) 

N_CHARS = len(char_names)

characters = pd.DataFrame({
    'character_name': char_names,
    'xp': np.random.randint(100, 10_000, N_CHARS),
    'gold': np.random.randint(0, 100_000, N_CHARS),
    'guild': np.random.choice(guild_names, N_CHARS)
})

characters['level'] = characters.xp // 350 + 1
characters.loc[np.random.choice(np.arange(N_CHARS), N_CHARS // 10), 'guild'] = np.nan
characters['character_id'] = characters.character_name.apply(to_id)
characters['guild_id'] = characters.guild.astype(str).apply(to_id)
characters['guild_id'] = np.where(characters['guild'].isna(), np.nan, characters['guild_id'])
characters['class'] = [', '.join(np.random.choice(class_names, np.random.randint(1, 3)).tolist()) for _ in range(N_CHARS)]

characters['user_id'] = np.append(
    np.random.choice(user_ids, N_USERS, replace=False),
    np.random.choice(user_ids, N_CHARS - N_USERS, replace=True),
)
```

### Join to OBT

```python
res = characters \
    .merge(users, how='outer', on='user_id') \
    .merge(payment_methods, on='user_id')
```
