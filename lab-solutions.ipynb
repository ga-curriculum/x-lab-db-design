#### Mart Table 1: Guild Analytics

Create a `guild_roster` table that includes each guild ID, guild name, and how many characters are in each guild.

````python
query = """
SELECT
    g.guild_id,
    g.guild,
    COUNT(c.character_name) AS n_members
FROM guilds AS g
LEFT JOIN characters AS c
ON c.guild_id = g.guild_id
GROUP BY g.guild_id, g.guild
"""

guild_roster = pd.read_sql(query, conn)
guild_roster.head()
````

#### Mart Table 2: Characters Cleaned

Create a version of the `characters` table denormalized to include:
* The user's full name (one field!)
* The user's email address
* The character's guild

````python
query = """
SELECT
    c.*,
    u.first || " " || u.last AS full_name,
    u.email,
    g.guild_id,
    g.guild
FROM characters AS c
LEFT JOIN users AS u
ON u.user_id = c.user_id
LEFT JOIN guilds AS g
ON g.guild_id = c.guild_id
"""

character_cleaned = pd.read_sql(query, conn)
character_cleaned.head()
````

#### Mart Table 3: Billing Update

Suppose billing rules are going to change for credit cards expiring in or after 2030. Create a table of all the characters (just name and ID is fine) whose users have a credit card expiring in or after 2030, as the characters will need to be notified when doing in-app purchases in-game.

_Note:_ This one is a little tough! Feel free to use ChatGPT.

````python
query = """
WITH cc_exp AS (
    SELECT
        user_id,
        CAST(SUBSTR(cc_expire, -2) AS INTEGER) AS year
    FROM payment_methods
)
SELECT
    c.character_id,
    c.character_name,
    e.user_id,
    e.year AS cc_expire_year
FROM characters AS c
LEFT JOIN cc_exp AS e
ON e.user_id = c.user_id
WHERE year >= 30
"""

billing_update = pd.read_sql(query, conn)
billing_update.head()
````

#### Mart Table 4: Guild Data

Suppose the BI analysts want to gather guild data alongside character information. Create a mart table that is the character table plus a new column that is the number of guildmates they have. Note that the number of of guildmates is the total number of characters in a character's guild minus one (since they do not count themselves). Make sure to handle the case where a character is not in a guild at all.

````python
query = """
WITH guild_counts AS (
    SELECT
        g.guild_id,
        COUNT(c.character_id) AS n_members
    FROM characters AS c
    LEFT JOIN guilds AS g
    ON g.guild_id = c.guild_id
    WHERE g.guild_id NOT NULL
    GROUP BY g.guild_id
)
SELECT
    c.*,
    CASE
        WHEN gc.n_members IS NULL THEN 0
        ELSE gc.n_members - 1
    END AS n_guildmates
FROM characters AS c
LEFT JOIN guild_counts AS gc
ON c.guild_id = gc.guild_id
"""

guild_data = pd.read_sql(query, conn)
guild_data
````
